```python
from openai import OpenAI
client = OpenAI(base_url="http://localhost:11434/v1", api_key="unused")
```

```python

emb = client.embeddings.create(
    model="octen-embedding-0.6b",
    input="hello world",
)
print(emb.data[0].embedding[:8])
```

```python
with client.audio.speech.with_streaming_response.create(
    model="outetts", voice="af_sky", input="Hello from llamaswap",
) as resp:
    resp.stream_to_file("out.wav")
```

```python
with open("data/script_scarlette_ref.wav", "rb") as fh:
    tr = client.audio.transcriptions.create(
        model="whisper-asr", file=fh, response_format="json",
    )
    print(tr.text)
```

```python
import base64
img = client.images.generate(
    model="qwen-image-2.1-uc",
    prompt="a cat piloting a tiny spaceship",
    size="1024x1024",
)
open("cat.png", "wb").write(base64.b64decode(img.data[0].b64_json))
```

```python
import requests
r = requests.post("http://localhost:11434/v1/audio/music", json={
    "model": "minimax-music3",
    "input": "A bright pop rock song with clean drums and a clear male vocal.",
    "lyrics": "[verse] City lights are shining low.\n[chorus] Turn it up and let it fly.",
    "duration": 30,
})
open("song.wav", "wb").write(r.content)
```
